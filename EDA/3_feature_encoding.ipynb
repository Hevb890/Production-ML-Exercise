### 1. Import Dependencies

```python
import os
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("./data/processed/ChurnModeling_Binning_Applied.csv")
df.head()
```

### 3. Encode Nominal Variables

```python
geography_dummies = pd.get_dummies(df['Geography'], prefix='Geogeraphy')
geography_dummies
```

```python
gender_dummies = pd.get_dummies(df['Gender'], prefix='Gender')
gender_dummies
```

```python
df_encoded = pd.concat([df, geography_dummies], axis=1)
del df_encoded['Geography']
```

```python
df_encoded
```

```python
df_encoded = pd.concat([df_encoded, gender_dummies], axis=1)
del df_encoded['Gender']
```

```python
df_encoded
```

### 2. Encode Ordinal Variables

```python
encode_dict_creditscore = {
    'Poor': 0,
    'Fair': 1,
    'Good': 2,
    'Very Good': 3,
    'Excellent': 4
}

df_encoded['CreditScoreBins'] = df_encoded['CreditScoreBins'].map(encode_dict_creditscore)
df_encoded.head()
```

```python
df_encoded.to_csv("./data/processed/ChurnModeling_Feature_Encoding_Applied.csv", index=False)
```
